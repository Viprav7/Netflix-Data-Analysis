```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_fixed.csv")
df.head()
```

```python


yearly = df.groupby(['Year_added', 'Type']).size().unstack()

fig, ax = plt.subplots(figsize=(12, 5), facecolor='#141414')
ax.set_facecolor('#141414')

yearly.plot(kind='bar', stacked=True, color=['#E50914', '#564d4d'], ax=ax)

ax.set_title('Netflix Content Added Per Year', color='white', fontsize=14, pad=15)
ax.set_xlabel('Year', color='white')
ax.set_ylabel('Number of Titles', color='white')
ax.tick_params(colors='white')
ax.xaxis.set_tick_params(rotation=0)

legend = ax.get_legend()
legend.get_frame().set_facecolor('#221F1F')
for text in legend.get_texts():
    text.set_color('white')

plt.tight_layout()
plt.savefig('content_growth.png', dpi=150, facecolor='#141414')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np

BG       = '#141414'
RED      = '#E50914'
DARK_RED = '#564d4d'
WHITE    = '#FFFFFF'
GRAY     = '#AAAAAA'

def apply_dark_theme(ax, title):
    ax.set_facecolor(BG)
    ax.set_title(title, color=WHITE, fontsize=14, pad=15, fontweight='bold')
    ax.tick_params(colors=WHITE)
    ax.xaxis.label.set_color(WHITE)
    ax.yaxis.label.set_color(WHITE)
    for spine in ax.spines.values():
        spine.set_edgecolor(GRAY)
    ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{int(x):,}'))

fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)

yearly = df.groupby(['Year_added', 'Type']).size().unstack()
yearly.plot(kind='bar', stacked=True, color=[RED, DARK_RED], ax=ax)

apply_dark_theme(ax, 'Netflix Content Added Per Year')
ax.set_xlabel('Year', color=WHITE)
ax.set_ylabel('Number of Titles', color=WHITE)
ax.xaxis.set_tick_params(rotation=0)


peak_year_idx = yearly.sum(axis=1).argmax()
peak_val = yearly.sum(axis=1).max()
ax.annotate(f'Peak: {int(peak_val)} titles',
            xy=(peak_year_idx, peak_val),
            xytext=(peak_year_idx - 2, peak_val - 300),
            arrowprops=dict(arrowstyle='->', color=WHITE),
            color=WHITE, fontsize=9)

legend = ax.get_legend()
legend.get_frame().set_facecolor('#221F1F')
for text in legend.get_texts():
    text.set_color(WHITE)

plt.tight_layout()
plt.savefig('chart1_content_growth.png', dpi=150, facecolor=BG)
plt.show()
print("✅ Chart 1 saved")


fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)

top_genres = df['Primary_genre'].value_counts().head(10)


bar_colors = [RED if i == 0 else DARK_RED for i in range(len(top_genres))]
bars = ax.bar(top_genres.index, top_genres.values, color=bar_colors, edgecolor='none')


for bar in bars:
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 15,
            f'{int(bar.get_height()):,}', ha='center', color=WHITE, fontsize=8)

apply_dark_theme(ax, 'Top 10 Genres on Netflix')
ax.set_xlabel('Genre', color=WHITE)
ax.set_ylabel('Number of Titles', color=WHITE)
ax.set_xticks(range(len(top_genres)))
ax.set_xticklabels(top_genres.index, rotation=35, ha='right', color=WHITE)

plt.tight_layout()
plt.savefig('chart3_top_genres.png', dpi=150, facecolor=BG)
plt.show()
print("✅ Chart 3 saved")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(14, 8), facecolor=BG)


top_genres  = df['Primary_genre'].value_counts().head(10).index
top_ratings = df['Rating'].value_counts().head(8).index

heatmap_data = (df[df['Primary_genre'].isin(top_genres) & df['Rating'].isin(top_ratings)]
                .groupby(['Primary_genre', 'Rating'])
                .size()
                .unstack(fill_value=0))


rating_order = ['TV-Y', 'TV-G', 'PG', 'TV-PG', 'PG-13', 'TV-14', 'TV-MA', 'R']
rating_order = [r for r in rating_order if r in heatmap_data.columns]
heatmap_data = heatmap_data[rating_order]


sns.heatmap(
    heatmap_data,
    ax=ax,
    cmap='Reds',
    linewidths=0.5,
    linecolor='#222222',
    annot=True,
    fmt='d',
    annot_kws={'color': WHITE, 'fontsize': 9},
    cbar_kws={'shrink': 0.8}
)


cbar = ax.collections[0].colorbar
cbar.ax.yaxis.set_tick_params(color=WHITE)
plt.setp(cbar.ax.yaxis.get_ticklabels(), color=WHITE)
cbar.set_label('Number of Titles', color=WHITE)


ax.set_facecolor(BG)
fig.patch.set_facecolor(BG)
ax.set_title('Rating Distribution Across Top 10 Genres', color=WHITE, fontsize=14,
             fontweight='bold', pad=20)
ax.set_xlabel('Content Rating', color=WHITE, fontsize=11)
ax.set_ylabel('Genre', color=WHITE, fontsize=11)
plt.setp(ax.get_xticklabels(), color=WHITE, fontsize=10)
plt.setp(ax.get_yticklabels(), color=WHITE, fontsize=10, rotation=0)

plt.tight_layout()
plt.savefig('chart2_rating_heatmap.png', dpi=150, facecolor=BG)
plt.show()
print("Heatmap saved")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(14, 7), facecolor=BG)
ax.set_facecolor(BG)


movies = df[(df['Type'] == 'MOVIE') & (df['Movies_duration'].notna())]


rating_order = (movies.groupby('Rating')['Movies_duration']
                .median()
                .sort_values()
                .index.tolist())

sns.boxplot(
    data=movies,
    x='Rating',
    y='Movies_duration',
    order=rating_order,
    color=DARK_RED,
    medianprops=dict(color=RED, linewidth=2.5),
    whiskerprops=dict(color=WHITE, linewidth=1.2),
    capprops=dict(color=WHITE, linewidth=1.2),
    boxprops=dict(edgecolor=WHITE, linewidth=1),
    flierprops=dict(marker='o', color=RED, alpha=0.4, markersize=4),
    ax=ax
)


ax.set_facecolor(BG)
fig.patch.set_facecolor(BG)
ax.set_title('Movie Duration Distribution by Rating', color=WHITE,
             fontsize=14, fontweight='bold', pad=20)
ax.set_xlabel('Content Rating', color=WHITE, fontsize=11)
ax.set_ylabel('Duration (minutes)', color=WHITE, fontsize=11)
plt.setp(ax.get_xticklabels(), color=WHITE, fontsize=10)
plt.setp(ax.get_yticklabels(), color=WHITE, fontsize=10)
for spine in ax.spines.values():
    spine.set_edgecolor('#444444')


medians = movies.groupby('Rating')['Movies_duration'].median()
for i, rating in enumerate(rating_order):
    ax.text(i, medians[rating] + 4, f'{int(medians[rating])}m',
            ha='center', color=RED, fontsize=8, fontweight='bold')


ax.axhline(y=90, color=WHITE, linestyle='--', linewidth=1, alpha=0.4)
ax.text(len(rating_order) - 0.5, 92, '90 min mark',
        color=WHITE, fontsize=8, alpha=0.6, ha='right')

plt.tight_layout()
plt.savefig('chart_boxplot_duration.png', dpi=150, facecolor=BG)
plt.show()
print("Box plot saved")
```

```python
fig, ax = plt.subplots(figsize=(13, 6), facecolor=BG)
ax.set_facecolor(BG)

movies = df[(df['Type'] == 'MOVIE') & (df['Movies_duration'].notna())]


n, bins, patches = ax.hist(movies['Movies_duration'], bins=40,
                            color=DARK_RED, edgecolor=BG, linewidth=0.5)


for patch, left_edge in zip(patches, bins[:-1]):
    if 90 <= left_edge <= 120:
        patch.set_facecolor(RED)


mean_val   = movies['Movies_duration'].mean()
median_val = movies['Movies_duration'].median()

ax.axvline(mean_val,   color='#FFD700', linestyle='--', linewidth=1.8, label=f'Mean: {mean_val:.0f} min')
ax.axvline(median_val, color='#00CED1', linestyle='--', linewidth=1.8, label=f'Median: {median_val:.0f} min')


ax.text(mean_val + 2,   n.max() * 0.95, f'{mean_val:.0f}m',   color='#FFD700', fontsize=9)
ax.text(median_val + 2, n.max() * 0.85, f'{median_val:.0f}m', color='#00CED1', fontsize=9)


fig.patch.set_facecolor(BG)
ax.set_title('Movie Duration Distribution on Netflix', color=WHITE,
             fontsize=14, fontweight='bold', pad=20)
ax.set_xlabel('Duration (minutes)', color=WHITE, fontsize=11)
ax.set_ylabel('Number of Movies',   color=WHITE, fontsize=11)
plt.setp(ax.get_xticklabels(), color=WHITE, fontsize=10)
plt.setp(ax.get_yticklabels(), color=WHITE, fontsize=10)
for spine in ax.spines.values():
    spine.set_edgecolor('#444444')


legend = ax.legend(facecolor='#221F1F', labelcolor=WHITE, fontsize=10)



plt.tight_layout()
plt.savefig('chart_duration_histogram.png', dpi=150, facecolor=BG)
plt.show()
print("Histogram saved")
```
